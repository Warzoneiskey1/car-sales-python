# Car Sales Analysis – Jeffrey Godwin

Cleaned and analyzed 8,000+ rows of car sales data using Python Pandas.
Found top-selling models and seasonal trends.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data (you can change the file name later)
df = pd.read_csv('car_sales.csv')
print('Rows:', len(df))
df.head()
```

```python
# Cleaning
df.drop_duplicates(inplace=True)
df['Revenue'] = df['Quantity'] * df['Price']
print('Clean rows:', len(df))
```

```python
# Top 10 models
top_models = df.groupby('Model')['Quantity'].sum().sort_values(ascending=False).head(10)
print(top_models)

plt.figure(figsize=(10,6))
sns.barplot(x=top_models.values, y=top_models.index)
plt.title('Top 10 Car Models by Sales')
plt.xlabel('Units Sold')
plt.show()
```
